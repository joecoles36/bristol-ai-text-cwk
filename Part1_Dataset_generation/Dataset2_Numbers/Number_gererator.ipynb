```python
import random
import pandas as pd
```

```python
# Vocabulary
colours = ['red', 'blue', 'green', 'yellow', 'orange', 
           'purple', 'pink', 'brown', 'grey']

sizes = ['small', 'medium', 'large']

# Numbers by digit length
one_digit = list(range(1, 10))        # 1-9
two_digit = list(range(10, 100))      # 10-99
three_digit = list(range(100, 1000))  # 100-999
four_digit = list(range(1000, 10000)) # 1000-9999
```

```python
def get_article(word):
    if word[0].lower() in ['a', 'e', 'i', 'o', 'u']:
        return 'an'
    return 'a'
```

```python
# Sentence templates
templates = [
    "a {size} {colour} {number}",
    "the number {number} in {colour} colour and {size} size",
    "{article} {colour} number {number} of {size} size",
    "a {size} {number} in {colour} colour",
    "a {size} {colour} number {number}"
]
```

```python
def generate_sentence(template, number, colour, size):
    article = get_article(colour)
    return template.format(
        article=article,
        number=number,
        colour=colour,
        size=size
    )

# Generating sentences
sentences = []
sentence_id = 1

for i in range(4000):
    template = random.choice(templates)
    colour = random.choice(colours)
    size = random.choice(sizes)

    # Pick a number of random digit length
    digit_length = random.choice([1, 2, 3, 4])
    if digit_length == 1:
        number = random.choice(one_digit)
    elif digit_length == 2:
        number = random.choice(two_digit)
    elif digit_length == 3:
        number = random.choice(three_digit)
    else:
        number = random.choice(four_digit)

    sentence = generate_sentence(template, number, colour, size)

    sentences.append({
        'sentence_id': sentence_id,
        'sentence': sentence,
        'number': number,
        'num_digits': digit_length,
        'colour': colour,
        'size': size,
        'template_id': templates.index(template) + 1
    })
    
    sentence_id += 1

print(f"Generated {len(sentences)} sentences")
print("\nExample sentences:")
for s in random.sample(sentences, 20):
    print(s['sentence'])
```

```python
# Check duplicates
df = pd.DataFrame(sentences)
duplicates = df[df.duplicated(subset=['sentence'])]
print(f"Duplicate sentences: {len(duplicates)}")
```

```python
# Remove duplicates
df = df.drop_duplicates(subset=['sentence']).reset_index(drop=True)
df['sentence_id'] = range(1, len(df) + 1)
```

```python
print(f"Cleaned dataset: {len(df)} sentences")
print("\nTemplate distribution:")
print(df['template_id'].value_counts().sort_index())
print("\nDigit length distribution:")
print(df['num_digits'].value_counts().sort_index())
```

```python
# Saving to csv file
df.to_csv('sentence_dataset_numbers.csv', index=False)
print("\nSaved to sentence_dataset_numbers.csv")
```
