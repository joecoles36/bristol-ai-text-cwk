```python
import os
import warnings
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer

# Environment cleanup
os.environ['HF_HUB_DISABLE_SYMLINKS_WARNING'] = '1'
warnings.filterwarnings("ignore", message=".*unauthenticated requests.*")

# Initialize Model
print("Loading Embedding Model...")
model = SentenceTransformer('all-MiniLM-L6-v2')

def export_embeddings(csv_file, label_name):
    """
    Transforms raw CSV sentences into .npy binary label files.
    """
    try:
        df = pd.read_csv(csv_file)
        print(f"Processing {csv_file} ({len(df)} rows)...")
        
        # Core Task: Mapping natural language to a numerical vector space
        embeddings = model.encode(df['sentence'].tolist())
        
        # Save as .npy for direct integration into Person 4's training pipeline
        np.save(f'{label_name}_labels.npy', embeddings)
        print(f"Export Success: {label_name}_labels.npy saved with shape {embeddings.shape}\n")
    except FileNotFoundError:
        print(f"Skip: {csv_file} not found.\n")

# Generate final targets for all three sub-tasks
export_embeddings('sentence_dataset_shapes.csv', 'shapes')
export_embeddings('sentence_dataset_numbers.csv', 'numbers')
export_embeddings('sentence_dataset_tictactoe.csv', 'tictactoe')

print("--- All Text Labels Ready for Neural Network Training ---")
```
